```python
# ATM AND UPI TRANSACTION SIMULATOR

# INITIAL DETAILS
correct_pin = "1234"
balance = 5000.00
transactions = []

# STEP 1: LOGIN
print("\n==================================================")
print("          ATM & UPI TRANSACTION SIMULATOR")
print("==================================================")

attempts = 3
login_successful = False

while attempts > 0:
    pin = input("\nEnter your 4-digit PIN: ")

    if pin == correct_pin:
        print("\n--------------------------------------------------")
        print("                 LOGIN SUCCESSFUL!")
        print("--------------------------------------------------")
        login_successful = True
        break

    else:
        attempts = attempts - 1

        if attempts > 0:
            print("\nIncorrect PIN! Please try again.")
            print("Attempts remaining:", attempts)
        else:
            print("\n==================================================")
            print("           TOO MANY INCORRECT ATTEMPTS!")
            print("                 ACCESS DENIED")
            print("==================================================")


# STEP 2: MAIN MENU
if login_successful:

    while True:
        print("\n=================== MAIN MENU ===================")
        print("  1. Check Balance")
        print("  2. Deposit Money")
        print("  3. Withdraw Money")
        print("  4. Make UPI Payment")
        print("  5. Transaction History")
        print("  6. Exit")
        print("--------------------------------------------------")

        choice = input("Enter your choice: ")

        # OPTION 1: CHECK BALANCE
        if choice == "1":
            print("\n---------------- CHECK BALANCE ------------------")
            print("Available Balance : Rs. {:.2f}".format(balance))
            print("--------------------------------------------------")

        # OPTION 2: DEPOSIT MONEY
        elif choice == "2":
            print("\n---------------- DEPOSIT MONEY ------------------")

            try:
                amount = float(input("Enter amount in Rs.: "))

                if amount > 0:
                    balance = balance + amount

                    transactions.append(
                        "Deposit: Rs. {:.2f}".format(amount)
                    )

                    print("\n             DEPOSIT SUCCESSFUL!")
                    print("--------------------------------------------------")
                    print("Amount Deposited  : Rs. {:.2f}".format(amount))
                    print("Updated Balance   : Rs. {:.2f}".format(balance))
                    print("--------------------------------------------------")

                else:
                    print("Enter an amount greater than zero.")

            except ValueError:
                print("Invalid amount! Please enter a number.")

        # OPTION 3: WITHDRAW MONEY
        elif choice == "3":
            print("\n---------------- WITHDRAW MONEY -----------------")

            try:
                amount = float(input("Enter amount in Rs.: "))

                if amount <= 0:
                    print("Enter an amount greater than zero.")

                elif amount > balance:
                    transactions.append(
                        "FAILED Withdrawal: Rs. {:.2f} - Insufficient balance".format(
                            amount
                        )
                    )

                    print("\n          WITHDRAWAL FAILED!")
                    print("Reason: Insufficient balance.")
                    print("Available Balance : Rs. {:.2f}".format(balance))
                    print("--------------------------------------------------")

                else:
                    balance = balance - amount

                    transactions.append(
                        "Withdrawal: Rs. {:.2f}".format(amount)
                    )

                    print("\n           WITHDRAWAL SUCCESSFUL!")
                    print("--------------------------------------------------")
                    print("Amount Withdrawn  : Rs. {:.2f}".format(amount))
                    print("Updated Balance   : Rs. {:.2f}".format(balance))
                    print("--------------------------------------------------")

            except ValueError:
                print("Invalid amount! Please enter a number.")

        # OPTION 4: UPI PAYMENT
        elif choice == "4":
            print("\n---------------- UPI PAYMENT --------------------")

            upi_id = input("Enter receiver's UPI ID: ")

            if ("@" not in upi_id
                    or upi_id.startswith("@")
                    or upi_id.endswith("@")):

                transactions.append(
                    "FAILED UPI Payment - Invalid UPI ID: {}".format(upi_id)
                )

                print("Invalid UPI ID!")

            else:
                try:
                    amount = float(input("Enter amount in Rs.: "))

                    if amount <= 0:
                        print("Enter an amount greater than zero.")

                    elif amount > balance:
                        transactions.append(
                            "FAILED UPI Payment: Rs. {:.2f} to {} - Insufficient balance".format(
                                amount, upi_id
                            )
                        )

                        print("\n          PAYMENT FAILED!")
                        print("Reason: Insufficient balance.")
                        print("Available Balance : Rs. {:.2f}".format(balance))
                        print("--------------------------------------------------")

                    else:
                        balance = balance - amount

                        transactions.append(
                            "UPI Payment: Rs. {:.2f} to {}".format(
                                amount, upi_id
                            )
                        )

                        print("\n            PAYMENT SUCCESSFUL!")
                        print("--------------------------------------------------")
                        print("Paid To           : {}".format(upi_id))
                        print("Amount Paid       : Rs. {:.2f}".format(amount))
                        print("Updated Balance   : Rs. {:.2f}".format(balance))
                        print("--------------------------------------------------")

                except ValueError:
                    print("Invalid amount! Please enter a number.")

        # OPTION 5: TRANSACTION HISTORY
        elif choice == "5":
            print("\n============== TRANSACTION HISTORY ==============")

            if len(transactions) == 0:
                print("No transactions yet.")

            else:
                for transaction in transactions:
                    print("- " + transaction)

            print("==================================================")

        # OPTION 6: EXIT
        elif choice == "6":
            print("\n==================================================")
            print("       THANK YOU FOR USING OUR ATM SYSTEM!")
            print("==================================================")
            break

        # INVALID MENU CHOICE
        else:
            print("\nInvalid choice! Please select a number from 1 to 6.")
```
